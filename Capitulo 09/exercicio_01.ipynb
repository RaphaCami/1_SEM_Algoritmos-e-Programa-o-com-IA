```python
pontuacao = float(input('Digite a pontuação de risco (0 a 100): '))

if pontuacao <= 30:
    risco = 'Baixo risco'
elif pontuacao <= 70:
    risco = 'Risco médio'
else:
    risco = 'Alto risco'

print(f'Classificação: {risco}')
```
