```python
# ════════════════════════════════════════════════
# PROJETO 1 — Calculadora de Aprovação
# ════════════════════════════════════════════════

# ── COLETA ──────────────────────────────────────
print('=== Calculadora de Aprovação ===')
p1   = float(input('Nota Prova 1 (peso 30%): '))
p2   = float(input('Nota Prova 2 (peso 30%): '))
p3   = float(input('Nota Prova 3 (peso 20%): '))
proj = float(input('Nota Projetos (peso 20%): '))

# ── PROCESSAMENTO ───────────────────────────────
media = (p1 * 0.30) + (p2 * 0.30) + (p3 * 0.20) + (proj * 0.20)
diferenca = media - 7.0

# ── CLASSIFICAÇÃO ───────────────────────────────
if media >= 7:
    resultado = 'Aprovado'
elif media >= 5:
    resultado = 'Recuperação'
else:
    resultado = 'Reprovado'

# ── EXIBIÇÃO ────────────────────────────────────
print()
print('─' * 40)
print(' BOLETIM DO ALUNO ')
print('─' * 40)
print(f'Prova 1 (30%):   {p1:5.1f}')
print(f'Prova 2 (30%):   {p2:5.1f}')
print(f'Prova 3 (20%):   {p3:5.1f}')
print(f'Projetos (20%):  {proj:5.1f}')
print('─' * 40)
print(f'Média ponderada: {media:5.2f}')
print(f'Situação: {resultado}')
print()

if diferenca >= 0:
    print(f'Você está {diferenca:.2f} pontos ACIMA da aprovação.')
else:
    print(f'Você precisa de mais {abs(diferenca):.2f} pontos '
          f'para aprovação.')
print('─' * 40)
```
