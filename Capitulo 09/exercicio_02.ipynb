**Exercicio 1**

```python
nome    = input('Nome do vendedor: ')
meta    = float(input('Meta de vendas: R$ '))
vendido = float(input('Total vendido: R$ '))

diferenca = vendido - meta

if vendido > meta:
    classificacao = 'Acima da meta'
elif vendido == meta:
    classificacao = 'Na meta'
else:
    classificacao = 'Abaixo da meta'

print('-' * 40)
print('RELATÓRIO DE DESEMPENHO')
print('-' * 40)
print(f'Vendedor: {nome}')
print(f'Meta: R$ {meta:.2f}')
print(f'Total vendido: R$ {vendido:.2f}')
print(f'Diferença: R$ {diferenca:.2f}')
print(f'Classificação: {classificacao}')
print('-' * 40)
```
