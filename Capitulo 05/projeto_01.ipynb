```python
pago = input('O pedido foi pago? (s/n): ') == 's'

if not pago:
    print('Envie mensagem de cobrança')
else:
    em_estoque = input('Produto em estoque? (s/n): ') == 's'
    if em_estoque:
        print('Envie o pedido')
    else:
        print('Avise o cliente: indisponível')
```
