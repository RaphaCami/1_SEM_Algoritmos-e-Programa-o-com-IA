**Exercicio 2**

```python
soma = 0
contador = 0

nota = float(input('Nota (-1 para parar): '))
while nota != -1:
    soma = soma + nota
    contador = contador + 1
    nota = float(input('Nota (-1 para parar): '))

if contador > 0:
    print('Média:', soma / contador)
else:
    print('Nenhuma nota informada')
```
