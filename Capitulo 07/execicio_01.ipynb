```python
# COLETA
# RA pode ter letras ou zeros à esquerda: fica como str
nome        = input('Nome completo: ')
ra          = input('RA: ')
cidade      = input('Cidade: ')
semestre    = int(input('Semestre: '))
media_atual = float(input('Média atual: '))

# EXIBIÇÃO
print('==============================')
print('   CADASTRO DO ESTUDANTE')
print('==============================')
print(f'Nome:     {nome}')
print(f'RA:       {ra}')
print(f'Cidade:   {cidade}')
print(f'Semestre: {semestre}')
print(f'Média:    {media_atual:.2f}')
print('==============================')
```
