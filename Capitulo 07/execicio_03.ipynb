```python
# COLETA
investimento = float(input('Investimento em marketing (R$): '))
leads        = int(input('Leads gerados: '))
vendas       = int(input('Vendas realizadas: '))

# PROCESSAMENTO
custo_por_lead = investimento / leads
taxa_conversao = vendas / leads    # número entre 0 e 1

# EXIBIÇÃO
print('RELATÓRIO DA CAMPANHA')
print(f'Investimento:      R$ {investimento:.2f}')
print(f'Leads gerados:     {leads}')
print(f'Vendas realizadas: {vendas}')
print(f'Custo por lead:    R$ {custo_por_lead:.2f}')
# :.1% multiplica por 100 e acrescenta o símbolo %
print(f'Taxa de conversão: {taxa_conversao:.1%}')
```
