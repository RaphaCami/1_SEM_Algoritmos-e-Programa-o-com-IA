```python
# ============================================
# Calculadora de tamanho bruto de imagem RGB
# Altere largura e altura para testar outras resoluções
# ============================================

largura         = 1920   # pixels
altura          = 1080   # pixels
bytes_por_pixel = 3      # RGB: 1 byte por canal (R, G, B)

# Cálculo
tamanho_bytes = largura * altura * bytes_por_pixel
tamanho_kb    = tamanho_bytes / 1024
tamanho_mb    = tamanho_kb / 1024

# Resultado
print(f'Imagem:  {largura} x {altura} pixels')
print(f'Tamanho: {tamanho_bytes:,} bytes')
print(f'Tamanho: {tamanho_kb:,.2f} KB')
print(f'Tamanho: {tamanho_mb:.2f} MB')
```
