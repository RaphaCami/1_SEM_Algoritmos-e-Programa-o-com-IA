**Exercicio 5**

```python
numero = 42
restos = []

while numero > 0:
    print(f'{numero:2} ÷ 2 = {numero // 2:2} -> resto {numero % 2}')
    restos.append(numero % 2)
    numero = numero // 2

# Os restos são lidos de baixo para cima
binario = ''.join(str(r) for r in reversed(restos))
print('Lendo de baixo para cima:', binario)

# Verificação por potências de 2
print('Verificação: 32 + 8 + 2 =', 32 + 8 + 2)
print('Conferindo com format():', format(42, 'b'))
```
