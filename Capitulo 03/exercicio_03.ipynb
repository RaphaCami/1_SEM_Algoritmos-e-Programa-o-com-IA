```python
# Comparando o tamanho de três resoluções diferentes
resolucoes = [
    (640, 480, 'VGA'),
    (1920, 1080, 'Full HD'),
    (3840, 2160, '4K'),
]

for larg, alt, nome in resolucoes:
    tam = larg * alt * 3
    mb = tam / 1024 / 1024
    print(f'{nome:8s} {larg}x{alt:4d} -> {tam:,} bytes ({mb:.1f} MB)')
```
