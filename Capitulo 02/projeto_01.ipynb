**Projeto**

```python
palavra = 'RAPHAELLA'              # guarda o nome em uma variável

print('Letra | Decimal | Binário')   # cabeçalho da tabela
print('-' * 30)                       # linha separadora

for letra in palavra:                 # para cada letra...
    codigo = ord(letra)               # ...obtém o código decimal
    binario = format(codigo, '08b')   # binário com 8 dígitos
    print(f'{letra:5} | {codigo:7} | {binario}')
```
