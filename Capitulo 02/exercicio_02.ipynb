```python
# Confirmação com chr(): 67 -> C, 84 -> T, 78 -> N
print(chr(67))  # C
print(chr(84))  # T
print(chr(78))  # N

# Bônus: o Python converte o binário em decimal e em letra
for binario in ['01000011', '01010100', '01001110']:
    decimal = int(binario, 2)
    print(binario, '->', decimal, '->', chr(decimal))
```
