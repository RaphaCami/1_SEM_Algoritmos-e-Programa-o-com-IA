```python
valor_compra = float(input('Valor da compra: '))

if valor_compra >= 300:
    desconto = valor_compra * 0.10
else:
    desconto = 0

valor_final = valor_compra - desconto   # fora do if/else: vale para os dois caminhos

print('Desconto: R$', desconto)
print('Valor final: R$', valor_final)
```
