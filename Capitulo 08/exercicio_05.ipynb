```python
valor_pedido = float(input('Valor do pedido (R$): '))
cliente_recorrente = input('Cliente recorrente? (s/n): ')

# Cálculos aritméticos
imposto = valor_pedido * 0.12
total = valor_pedido + imposto

# Operador relacional: == compara a string digitada
recorrente = cliente_recorrente == 's'

# Operador lógico: or combina duas situações de aprovação
aprovado = valor_pedido > 300 or (recorrente and valor_pedido > 150)

print(f'Imposto (12%): R$ {imposto:.2f}')
print(f'Total com imposto: R$ {total:.2f}')
print(f'Aprovação automática: {aprovado}')
```
