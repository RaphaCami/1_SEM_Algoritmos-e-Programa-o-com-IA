```python
# Versão 1 — depende da precedência de and sobre or
# vip = valor_compra > 500 or anos_cliente > 2 and valor_compra > 200

# Versão 2 — parênteses explícitos
# vip = valor_compra > 500 or (anos_cliente > 2 and valor_compra > 200)

testes = [(600, 1), (100, 5), (300, 3)]

for valor_compra, anos_cliente in testes:
    versao1 = valor_compra > 500 or anos_cliente > 2 and valor_compra > 200
    versao2 = valor_compra > 500 or (anos_cliente > 2 and valor_compra > 200)
    print(f'valor={valor_compra}, anos={anos_cliente}')
    print(f'  Versão 1 (com precedência): {versao1}')
    print(f'  Versão 2 (com parênteses):  {versao2}')
```
