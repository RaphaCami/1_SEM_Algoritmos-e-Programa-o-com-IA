```python
preco_venda = float(input('Preço de venda (R$): '))
custo       = float(input('Custo (R$): '))

if preco_venda <= 0:
    print('Erro: o preço de venda precisa ser maior que zero.')
else:
    margem = (preco_venda - custo) / preco_venda * 100

    if margem < 10:
        classificacao = 'baixa'
    elif margem <= 30:
        classificacao = 'adequada'
    else:
        classificacao = 'alta'

    print(f'Margem: {margem:.2f}%')
    print(f'Classificação: {classificacao}'),
```
