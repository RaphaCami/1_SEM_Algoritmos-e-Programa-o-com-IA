```python
# ── COLETA DE DADOS ─────────────────────────────────
vendas = float(input('Total de vendas no mês (R$): '))
anos_empresa = int(input('Anos de empresa: '))

# ── VALIDAÇÃO ────────────────────────────────────────
# Rejeitamos valores negativos antes de qualquer cálculo
if vendas < 0 or anos_empresa < 0:
    print('Erro: vendas e anos de empresa não podem ser negativos.')
else:
    # ── PROCESSAMENTO ────────────────────────────────
    salario_base = 2000.0

    # Operador aritmético: * calcula 5% das vendas
    comissao = vendas * 0.05

    # As duas condições precisam ser verdadeiras para o bônus
    if vendas > 10000 and anos_empresa >= 1:
        bonus = 500.0
    else:
        bonus = 0.0

    # Operador aritmético: + soma os três componentes
    total = salario_base + comissao + bonus

    # ── SAÍDA FORMATADA ──────────────────────────────
    print('\n--- Demonstrativo de Pagamento ---')
    print(f'Salário base:  R$ {salario_base:>10,.2f}')
    print(f'Comissão (5%): R$ {comissao:>10,.2f}')
    print(f'Bônus:         R$ {bonus:>10,.2f}')
    print('-' * 34)
    print(f'TOTAL:         R$ {total:>10,.2f}')
```
