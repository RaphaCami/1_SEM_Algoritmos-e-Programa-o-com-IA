```python
valor_compra = float(input('Valor da compra (R$): '))
distancia    = float(input('Distância (km): '))

if valor_compra > 300 and distancia <= 20:
    frete = 0.0
else:
    frete = valor_compra * 0.08 + distancia * 1.50

print(f'Frete: R$ {frete:.2f}')
```
