```python
total = float(input('Total da compra: R$ '))

if total <= 100:
    porte = 'Compra pequena'
elif total <= 500:
    porte = 'Compra média'
else:
    porte = 'Compra grande'

print(f'Total: R$ {total:.2f}')
print(f'Classificação: {porte}')
```
