```python
total_gasto = float(input('Total gasto no mês: R$ '))

if total_gasto <= 100:
    classificacao = 'Cliente eventual'
elif total_gasto <= 500:
    classificacao = 'Cliente regular'
else:
    classificacao = 'Cliente prioritário'

print(f'Classificação: {classificacao}')
```
