```python
print('Ambiguidades em "organize a lista de alunos por desempenho":')
print('1. O que significa "desempenho"? Média, frequência, participação?')
print('2. A ordem é crescente ou decrescente?')
print('3. O que fazer em caso de empate?')
print()

# Problema reescrito: ordenar por média (decrescente); empate -> maior frequência
alunos = [
    {'nome': 'Ana',   'ra': 1, 'media': 8.5, 'frequencia': 90},
    {'nome': 'Bruno', 'ra': 2, 'media': 9.0, 'frequencia': 80},
    {'nome': 'Carla', 'ra': 3, 'media': 8.5, 'frequencia': 95},
]

ordenados = sorted(alunos, key=lambda a: (-a['media'], -a['frequencia']))

for a in ordenados:
    print(f"{a['nome']:6} média {a['media']:.1f}  frequência {a['frequencia']}%")
```
