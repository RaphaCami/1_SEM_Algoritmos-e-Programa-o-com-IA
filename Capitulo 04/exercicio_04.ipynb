```python
def calcular_compra_com_desconto(valor_original, percentual_desconto):
    desconto = valor_original * percentual_desconto / 100   # passo 2
    valor_final = valor_original - desconto                 # passo 3
    return valor_final

# Casos de teste
print(calcular_compra_com_desconto(200, 10))  # esperado: 180.0
print(calcular_compra_com_desconto(50, 10))   # esperado: 45.0
```
